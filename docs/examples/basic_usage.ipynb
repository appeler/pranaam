# Basic usage

Pranaam returns calibrated name-pattern estimates. It does not establish a person's religion. Use it only for validated aggregate research, never to label individuals or make consequential decisions.

```python
from pranaam import estimate_muslim_name_pattern

english = estimate_muslim_name_pattern(["Shah Rukh Khan", "Amitabh Bachchan"], lang="eng")
english
```

`muslim_score` is calibrated from 0 to 1. `abstained` identifies uncertain estimates, and unsupported scripts receive no score.

```python
hindi = estimate_muslim_name_pattern(["शाहरुख खान", "अमिताभ बच्चन"], lang="hin")
hindi[["name_pattern_estimate", "muslim_score", "abstained", "model_revision"]]
```
